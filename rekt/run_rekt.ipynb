```python
import pandas as pd
import numpy as np

df = pd.read_csv("merged_student_question_history.csv")

# Count how many questions each student answered
question_counts = []
for _, row in df.iterrows():
    count = 0
    for i in range(1, 365):
        col = f"q{i}_id"
        if col in df.columns and pd.notna(row[col]):
            count += 1
        else:
            break  # Once we hit first NA, rest should be NA
    question_counts.append(count)

question_counts = np.array(question_counts)

print(f"Students: {len(question_counts)}")
print(f"Min questions answered: {question_counts.min()}")
print(f"Max questions answered: {question_counts.max()}")
print(f"Mean questions answered: {question_counts.mean():.1f}")
print(f"Median questions answered: {np.median(question_counts):.1f}")
print(f"\nDistribution:")
print(f"  0-50:   {(question_counts <= 50).sum()} students")
print(f"  51-100: {((question_counts > 50) & (question_counts <= 100)).sum()} students")
print(f"  101-150: {((question_counts > 100) & (question_counts <= 150)).sum()} students")
print(f"  151-200: {((question_counts > 150) & (question_counts <= 200)).sum()} students")
print(f"  201+:   {(question_counts > 200).sum()} students")
```
