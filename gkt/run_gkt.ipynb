# GKT Training Pipeline with DKT Graph Integration

This notebook executes the complete pipeline to train a GKT model using a static graph derived from a trained DKT model.

## Prerequisites
- The `dkt` folder must be in the parent directory (`../dkt`).
- The `merged_student_question_history.csv` (or equivalent) should be available for DKT preprocessing.
- The `merged_gkt.csv` (or equivalent) should be available for GKT preprocessing.

## Pipeline Steps
1. **DKT Preprocessing**: Prepare data for the DKT model.
2. **DKT Training**: Train the DKT model to learn question relationships.
3. **GKT Preprocessing**: Prepare data for the GKT model.
4. **GKT Training**: Train GKT, automatically generating the DKT graph using the trained DKT model.
